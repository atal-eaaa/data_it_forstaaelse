# Python Intall Party
Inden vi kan få data til at arbejde for os, skal værktøjskassen være på plads. I denne lektion installerer og konfigurerer vi de centrale værktøjer, som du vil bruge gennem hele uddannelsen: Python, Visual Studio Code, Git og de vigtigste Python-biblioteker. Målet er enkelt: Når dagen er omme, skal du have skreve de første Python-kodelinjer.
- Installation af værktøjer
- Kodeforståelse
- "Hello World"
- Input og output
- Variable og datatyper i Python

```python
import os
import platform
import shutil
import subprocess
import sys

print("Python installation")
print(f"Version: {sys.version}")
print(f"Executable: {sys.executable}")
print(f"Platform: {platform.platform()}")

print("\nEnvironment")
print(f"Current directory: {os.getcwd()}")
print(f"Virtual environment: {os.environ.get('VIRTUAL_ENV', 'None')}")

print("\nGit installation")
git_path = shutil.which("git")

if git_path:
    print(f"Git executable: {git_path}")
    result = subprocess.run(
        ["git", "--version"],
        capture_output=True,
        text=True,
        check=False
    )
    print(result.stdout.strip())
else:
    print("Git is not installed or is not available in PATH.")
```
